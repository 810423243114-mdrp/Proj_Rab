# Data Ingestion, Cleaning & Preprocessing with Pandas
This Jupyter Notebook demonstrates loading a messy retail dataset, cleaning missing values, handling outliers, fixing data types, and performing feature engineering.

```python
import pandas as pd
import numpy as np

# 1. DATA INGESTION
github_url = "https://raw.githubusercontent.com/810423243114-mdrp/Proj_Rab/main/messy_retail_sales.csv"

try:
    df = pd.read_csv(github_url)
    print("Dataset loaded successfully from GitHub!")
except Exception as e:
    print(f"Could not load from GitHub ({e}), loading local file...")
    df = pd.read_csv("messy_retail_sales.csv")

print("--- BEFORE CLEANING --- INFO:")
print(df.info())
df.head()
```

```python
# 2. DATA CLEANING & PREPROCESSING
# A. Handle Missing Values
df['Product'] = df['Product'].fillna('Unknown')
df['Unit_Price'] = df['Unit_Price'].fillna(df['Unit_Price'].median())
df['Cost_Price'] = df['Cost_Price'].fillna(df['Cost_Price'].median())

# B. Handle Outliers & Invalid Quantities
df['Quantity'] = pd.to_numeric(df['Quantity'], errors='coerce')
median_qty = df[df['Quantity'] > 0]['Quantity'].median()
df.loc[df['Quantity'] <= 0, 'Quantity'] = median_qty

# C. Handle Incorrect Data Types (Convert Date)
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')
default_date = pd.to_datetime('2025-01-01')
df['Date'] = df['Date'].fillna(default_date)
```

```python
# 3. FEATURE ENGINEERING
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df['Month_Name'] = df['Date'].dt.strftime('%B')

# Calculate Financial Metrics
df['Total_Revenue'] = df['Quantity'] * df['Unit_Price']
df['Total_Cost'] = df['Quantity'] * df['Cost_Price']
df['Total_Profit'] = df['Total_Revenue'] - df['Total_Cost']
df['Profit_Margin_%'] = (df['Total_Profit'] / df['Total_Revenue']) * 100

print("--- AFTER CLEANING & FEATURE ENGINEERING ---")
print(df.info())
df.head()
```

```python
# 4. EXPORT CLEANED DATA
df.to_csv("clean_retail_sales.csv", index=False)
print("Cleaned dataset successfully exported to 'clean_retail_sales.csv'!")
```
